```python
import mysql.connector as mypy
```

```python
con=mypy.connect(
    host='localhost',
    user='root',
    password='root',
    database='empdept'
)
```

```python
print('con is obtained ',con)
```

```python
if con.is_connected():
    print('connection successfull')
else:
    print('Sorry')
sql='select ename , sal , job from emp'
```

```python
cur=con.cursor()
print('cursor() helps to execute the required SQL before running SQL  ',cur)
cur.execute(sql)
print('cursor() helps to execute the required SQL after Running SQL ',cur)
result=cur.fetchall()
print(result, type(result))
```

```python
for a, b, c in result:
    print(f"{a} works in {b} and has a job of {c}")
```

```python

sql='select ename, sal,job  from emp order by sal'
cur=con.cursor()

cur.execute(sql)
result=cur.fetchall()
```

```python
for a, b,c in result:
    print(f"{a} earns Rs. {b} as a {c}.")
```
